```python
import bs4
from langchain_community.document_loaders import WebBaseLoader
from fpdf import FPDF

# Initialize the loader with more generic parse options to ensure content is captured
loader = WebBaseLoader(
    web_paths=URL_LIST,
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(['p', 'h1', 'h2', 'h3', 'h4', 'h5', 'h6'])
    ),
)

# Load the documents
documents = loader.load()

# Extract text from documents
text_content = ""
for doc in documents:
    soup = bs4.BeautifulSoup(doc.page_content, 'html.parser')
    for tag in soup.find_all(['p', 'h1', 'h2', 'h3', 'h4', 'h5', 'h6']):
        text_content += tag.get_text() + "\n\n"

# Function to write text to a PDF
class PDF(FPDF):
    def header(self):
        self.set_font('Arial', 'B', 12)
        self.cell(0, 10, 'Loyalist College Policies', 0, 1, 'C')
        self.ln(10)
    
    def footer(self):
        self.set_y(-15)
        self.set_font('Arial', 'I', 8)
        self.cell(0, 10, f'Page {self.page_no()}', 0, 0, 'C')

    def chapter_title(self, title):
        self.set_font('Arial', 'B', 12)
        self.cell(0, 10, title, 0, 1, 'L')
        self.ln(5)
    
    def chapter_body(self, body):
        self.set_font('Arial', '', 12)
        self.multi_cell(0, 10, body)
        self.ln()

# Create PDF object
pdf = PDF()
pdf.add_page()

# Add text to PDF
pdf.chapter_title('Policies and Codes of Conduct')
pdf.chapter_body(text_content)

# Save the PDF
pdf.output('Loyalist_College_Policies.pdf')

print("PDF created successfully!")
```

```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

# Define the web paths
web_paths = URL_LIST[0:6]

# Initialize the loader with more generic parse options to ensure content is captured
loader = WebBaseLoader(
    web_paths=web_paths,
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(['p', 'h1', 'h2', 'h3', 'h4', 'h5', 'h6'])
    ),
)

# Load the documents
blog_docs = loader.load()

# Check if content was extracted
for doc in blog_docs:
    print(f"Source: {doc.metadata['source']}")
    print(f"Content: {doc.page_content[:500]}...\n")  # Print first 500 characters for brevity
```

```python
whole_data = blog_docs + pdf_text
whole_data
```

```python
# Split
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=200)
```

```python
# Make splits
data_splits = text_splitter.split_documents(whole_data)
data_splits
```

```python
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.llms import Ollama
## Load Ollama LAMA3 LLM model
llamma2_llm=Ollama(model="llama2")
llamma2_llm
```

```python
from langchain.vectorstores import FAISS
db=FAISS.from_documents(data_splits,OllamaEmbeddings(model="mxbai-embed-large"))
#db=FAISS.from_documents(data_splits,OllamaEmbeddings(model="llama2"))
llama2_retriever=db.as_retriever()
llama2_retriever
```

```python
from langchain.prompts import PromptTemplate

prompt_template = """You are a helpful assistant for students at Loyalist College or LCIT or Loyalist College in Toronto or Loyalist college in belleville 
which is under TBC(Toronto Business College) and answer the questions directly.
In the answer try to provide as much text as possible from "response" section in the source document without making much changes.
If the answer is not found  kindly state "For this information, please contact academics or student support."

CONTEXT: {context}

QUESTION: {question}"""


PROMPT = PromptTemplate(
    template=prompt_template, input_variables=["context", "question"]
)
chain_type_kwargs = {"prompt": PROMPT}


from langchain.chains import RetrievalQA

llama2_chain = RetrievalQA.from_chain_type(llm=llamma2_llm,
                            chain_type="stuff",
                            retriever=llama2_retriever,
                            input_key="query",
                            return_source_documents=True,
                            chain_type_kwargs=chain_type_kwargs)
```

```python
from time import time
start_time = time()
result = llama2_chain('Can I smoke in the campus?')
print(result['result'])
end_time = time()
print(f"Time taken to execute : {end_time - start_time} sec")
```
